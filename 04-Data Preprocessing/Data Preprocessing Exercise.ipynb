## 2. Data trasformation  
    * if data is continuous = standard scalar, MinMaxScalar, Robust Scalar
    * if data is discrete = Label Encoder, one hot encoder

```python
weather_data_copy_01 = weather_data.copy()
```

```python
weather_data_copy_01
```

```python
from sklearn.preprocessing import LabelEncoder
le_encoder = LabelEncoder()
weather_data_copy_01["Weather"]= le_encoder.fit_transform(weather_data_copy_01["Weather"])
weather_data_copy_01
```

## How to choose between Label Encoder and One Hot Encoder?

## Input Feature;

Based on whether it is Parametric or Non Parametric model, we will choose between LE or OE.

    • For Parametric models - better to go with OHE
    • Foe Non-Parametric models always go with Label Encoding

## Output Feature:
    Always output feature, it must be label encoder
